```python
tolerance = 0.0001  # Convergence à 0.01 °C près
erreur_1=10

t_depart_consigne_1 = 350 # 90°C en Kelvin
t_retour_estime_1 = 305 # Hypothèse de départ (60°C)

i=0
while erreur_1 > tolerance :
    print(i)
    # 1. Calcul du delta T et du débit massique basé sur l'état actuel
    delta_t_1 = t_depart_consigne_1 - t_retour_estime_1
    cp_supply = fluid.get_heat_capacity(t_depart_consigne_1)
    cp_return = fluid.get_heat_capacity(t_retour_estime_1)
    M_dot_estimative=Heat/(((cp_supply+cp_return)/2)*delta_t_1)
    # 2. Mise à jour du flow control dans le réseau
    net.flow_control.at[0,'controlled_mdot_kg_per_s']= M_dot_estimative
    net.circ_pump_mass.at[0,'mdot_flow_kg_per_s']=M_dot_estimative
    print(M_dot_estimative)
    # 3. Exécution du calcul du réseau
    pp.pipeflow(net, mode="bidirectional")
    # 4. Lecture de la VRAIE température de retour qui arrive à l'usine
    t_retour_reel_1 = net.res_circ_pump_mass.loc[mass_flow, "t_from_k"]
    print(t_retour_reel_1)
    # 5. Calcul de l'erreur pour le critère d'arrêt
    erreur_1 = abs(t_retour_estime_1 - t_retour_reel_1)
    # 6. Mise à jour pour la prochaine itération
    t_retour_estime_1 = t_retour_reel_1
    i=i+1
```

```python
import numpy as np
import pandapipes as pp
from pandapipes.pf.pipeflow_setup import get_fluid

# =============================================================================
# PARAMETROS DE CONTROL DE LA PLANTA 2 (dos chaufferies que se mezclan)
# =============================================================================
T_a = 360.0          # K  salida de la chaufferie A (impuesta)
T_b = 350.0          # K  salida de la chaufferie B (impuesta)
T_mez_obj = 355.0    # K  temperatura de la MEZCLA que quieres imponer (T_b <= T_mez_obj <= T_a)
Q_planta2 = 500e3    # W  potencia total que entrega la planta 2 (OJO: en W, no kW)
T_ret_ini = 305.0    # K  estimacion inicial de la temperatura de retorno a la planta 2
tol = 1e-3           # K  tolerancia de convergencia
n_iter_max = 50

assert T_b <= T_mez_obj <= T_a, "La T de mezcla debe estar entre T_b y T_a"

# =============================================================================
# RED Y NUDOS
# =============================================================================
net = pp.create_empty_network(fluid="water")


def junc(name, p_bar, t_k):
    return pp.create_junction(net, pn_bar=p_bar, tfluid_k=t_k, name=name)


# Ida
j_fuente_ida = junc("Fuente_Ida", 5.0, 353.15)
j_fuente_2_ida = junc("Fuente_2_Ida (MEZCLA)", 5.0, 353.15)   # nudo de mezcla, SIN bomba
j_nodo_ida = {n: junc(f"Nodo{n}_ida", 5.0, 353.15) for n in (1, 2, 3, 4)}
j_cons_ida = {n: junc(f"Consumidor{n}_Ida", 4.8, 353.15) for n in range(1, 7)}

# Retorno
j_fuente_ret = junc("Fuente_Retorno", 1.5, 333.15)
j_fuente_2_ret = junc("Fuente_2_Retorno", 1.5, 333.15)
j_nodo_ret = {n: junc(f"Nodo{n}_ret", 5.0, 353.15) for n in (1, 2, 3, 4)}
j_cons_ret = {n: junc(f"Consumidor{n}_Retorno", 1.7, 333.15) for n in range(1, 7)}

# Salida de cada chaufferie de la planta 2 (nudos propios, uno por chaufferie)
j_p2a = junc("Planta2_chaufferie_A_ida", 5.0, T_a)
j_p2b = junc("Planta2_chaufferie_B_ida", 5.0, T_b)

# =============================================================================
# TUBERIAS
# =============================================================================
U_W = 0.35 / (np.pi * 0.15)


def tubo(j_from, j_to, L_km, name, adiabatico=False):
    return pp.create_pipe_from_parameters(
        net, from_junction=j_from, to_junction=j_to,
        length_km=L_km, inner_diameter_mm=150,
        u_w_per_m2k=0.0 if adiabatico else U_W,
        text_k=273.2, name=name, k_mm=0.1 * 1000,
    )


def tubo_ida_ret(j_ida_from, j_ida_to, j_ret_from, j_ret_to, L_km, name):
    """Crea la tuberia de ida (from->to) y la de retorno (sentido contrario)."""
    tubo(j_ida_from, j_ida_to, L_km, f"Tubo_Ida_{name}")
    tubo(j_ret_to, j_ret_from, L_km, f"Tubo_Retorno_{name}")


# Planta 1 - nodo 1, Planta 2 (mezcla) - nodo 3
tubo_ida_ret(j_fuente_ida, j_nodo_ida[1], j_fuente_ret, j_nodo_ret[1], 0.0374, "Plant1_nodo1")
tubo_ida_ret(j_fuente_2_ida, j_nodo_ida[3], j_fuente_2_ret, j_nodo_ret[3], 0.0265, "Plant2_nodo3")

# Nodo - nodo
tubo_ida_ret(j_nodo_ida[1], j_nodo_ida[2], j_nodo_ret[1], j_nodo_ret[2], 0.0254, "nodo1_nodo2")
tubo_ida_ret(j_nodo_ida[1], j_nodo_ida[3], j_nodo_ret[1], j_nodo_ret[3], 0.0316, "nodo1_nodo3")
tubo_ida_ret(j_nodo_ida[3], j_nodo_ida[4], j_nodo_ret[3], j_nodo_ret[4], 0.0498, "nodo3_nodo4")

# Nodo - consumidor
acometidas = {1: (2, 0.0155), 2: (2, 0.0122), 3: (2, 0.0161),
              4: (4, 0.0160), 5: (4, 0.0158), 6: (4, 0.0176)}
for c, (nodo, L) in acometidas.items():
    tubo_ida_ret(j_nodo_ida[nodo], j_cons_ida[c], j_nodo_ret[nodo], j_cons_ret[c], L,
                 f"nodo{nodo}_consm_{c}")

# Tubos cortos y ADIABATICOS de cada chaufferie al nudo de mezcla
# (sin perdidas, para que la T de mezcla sea exactamente la del balance de energia)
tubo(j_p2a, j_fuente_2_ida, 0.001, "Chaufferie_A_mezcla", adiabatico=True)
tubo(j_p2b, j_fuente_2_ida, 0.001, "Chaufferie_B_mezcla", adiabatico=True)

# =============================================================================
# CONSUMIDORES
# =============================================================================
for c in range(1, 7):
    pp.create_heat_consumer(
        net, from_junction=j_cons_ida[c], to_junction=j_cons_ret[c],
        qext_w=150000, treturn_k=305, name=f"Consumidor_{c}",
    )

# =============================================================================
# FUENTES
# =============================================================================
# Planta 1: presion constante
pp.create_circ_pump_const_pressure(
    net, flow_junction=j_fuente_ida, return_junction=j_fuente_ret,
    p_flow_bar=3, plift_bar=0.5, t_flow_k=360, name="plant_1",
)

# Planta 2: dos chaufferies, cada una con su bomba de caudal masico y su T impuesta.
# Ambas vuelven al mismo nudo de retorno y vierten en nudos distintos que se unen
# en j_fuente_2_ida (nudo de mezcla).
fluid = get_fluid(net)

x = (T_mez_obj - T_b) / (T_a - T_b)        # fraccion de caudal que pasa por la chaufferie A
cp0 = fluid.get_heat_capacity(0.5 * (T_mez_obj + T_ret_ini))
m_tot0 = Q_planta2 / (cp0 * (T_mez_obj - T_ret_ini))

pump_a = pp.create_circ_pump_const_mass_flow(
    net, return_junction=j_fuente_2_ret, flow_junction=j_p2a,
    mdot_flow_kg_per_s=x * m_tot0, t_flow_k=T_a, p_flow_bar=None, type="t",
    name="plant_2_chaufferie_A",
)
pump_b = pp.create_circ_pump_const_mass_flow(
    net, return_junction=j_fuente_2_ret, flow_junction=j_p2b,
    mdot_flow_kg_per_s=(1 - x) * m_tot0, t_flow_k=T_b, p_flow_bar=None, type="t",
    name="plant_2_chaufferie_B",
)

# =============================================================================
# BUCLE: ajusta (1) el caudal total segun Q y T_retorno y (2) el reparto x
#        para que la mezcla salga a T_mez_obj
# =============================================================================
pp.pipeflow(net, mode="bidirectional")

t_ret_est = T_ret_ini
for it in range(n_iter_max):
    # Caudal total de la planta 2 para entregar Q_planta2 con la mezcla objetivo
    cp = fluid.get_heat_capacity(0.5 * (T_mez_obj + t_ret_est))
    m_tot = Q_planta2 / (cp * (T_mez_obj - t_ret_est))

    # Reparto entre las dos chaufferies (evitamos caudal 0 en una bomba)
    x = float(np.clip(x, 0.01, 0.99))
    net.circ_pump_mass.at[pump_a, "mdot_flow_kg_per_s"] = x * m_tot
    net.circ_pump_mass.at[pump_b, "mdot_flow_kg_per_s"] = (1 - x) * m_tot

    pp.pipeflow(net, mode="bidirectional")

    T_mez = net.res_junction.at[j_fuente_2_ida, "t_k"]
    t_ret_real = net.res_circ_pump_mass.at[pump_a, "t_from_k"]   # mismo nudo de retorno para A y B

    err_mez = T_mez_obj - T_mez
    err_ret = abs(t_ret_est - t_ret_real)
    print(f"it {it:2d} | x={x:.4f} | m_tot={m_tot:.3f} kg/s | "
          f"T_mez={T_mez:.3f} K (obj {T_mez_obj}) | T_ret={t_ret_real:.3f} K")

    if abs(err_mez) < tol and err_ret < tol:
        break

    # Correccion: dT_mez/dx = (T_a - T_b)
    x = x + err_mez / (T_a - T_b)
    t_ret_est = t_ret_real
else:
    print("AVISO: no convergio en n_iter_max iteraciones")

# =============================================================================
# RESULTADOS DE LA PLANTA 2
# =============================================================================
m_a = net.res_circ_pump_mass.at[pump_a, "mdot_from_kg_per_s"]
m_b = net.res_circ_pump_mass.at[pump_b, "mdot_from_kg_per_s"]
cp_a = fluid.get_heat_capacity(T_a)
cp_b = fluid.get_heat_capacity(T_b)
Q_a = abs(m_a) * cp_a * (T_a - t_ret_real)
Q_b = abs(m_b) * cp_b * (T_b - t_ret_real)
print("\n--- Planta 2 ---")
print(f"T chaufferie A = {T_a} K | T chaufferie B = {T_b} K | T mezcla = {T_mez:.3f} K")
print(f"m_A = {abs(m_a):.3f} kg/s | m_B = {abs(m_b):.3f} kg/s")
print(f"Q_A = {Q_a/1e3:.1f} kW | Q_B = {Q_b/1e3:.1f} kW | total = {(Q_a+Q_b)/1e3:.1f} kW")

print(net.res_circ_pump_mass)
print(net.res_circ_pump_pressure)
print(net.res_pipe)
```
