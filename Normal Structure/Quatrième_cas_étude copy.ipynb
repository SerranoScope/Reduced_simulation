```python
import numpy as np
from pandapipes.component_models.abstract_models.circulation_pump import CirculationPump
from pandapipes.idx_branch import D, AREA, TOUTINIT
from pandapipes.idx_node import TINIT
from pandapipes.pf.pipeflow_setup import get_lookup

def _create_pit_branch_entries_fixed(cls, net, branch_pit):
    node_pit = net["_pit"]["node"]
    junction_idx_lookups = get_lookup(net, "node", "index")[
        cls.get_connected_node_type().table_name()]

    # FIX: usar la tabla COMPLETA, sin filtrar por in_service,
    # para que coincida con el tamaño de circ_pump_pit (que sí incluye
    # todas las filas, activas o no).
    circ_pump_tbl = net[cls.table_name()]
    circ_pump_pit = super(CirculationPump, cls).create_pit_branch_entries(net, branch_pit)
    circ_pump_pit[:, D] = 0.1
    circ_pump_pit[:, AREA] = circ_pump_pit[:, D] ** 2 * np.pi / 4

    types = circ_pump_tbl.type.values
    mask_t = np.isin(types, ["pt", "t"])
    juncts = circ_pump_tbl[cls.from_to_node_cols()[1]].values
    to_nodes = junction_idx_lookups[juncts]
    circ_pump_pit[mask_t, TOUTINIT] = circ_pump_tbl.t_flow_k.values[mask_t]
    circ_pump_pit[~mask_t, TOUTINIT] = node_pit[to_nodes[~mask_t], TINIT]
    return circ_pump_pit

CirculationPump.create_pit_branch_entries = classmethod(_create_pit_branch_entries_fixed)
```

```python
tolerance = 0.0000001  # Convergence à 0.01 °C près
erreur_1=10
erreur_2=10
t_depart_consigne_1 = 377 # 90°C en Kelvin
t_retour_estime_1 = 305 # Hypothèse de départ (60°C)
t_depart_consigne_2 = 376 # 90°C en Kelvin
t_retour_estime_2 = 305   # Hypothèse de départ (60°C)

i=0
while erreur_1 > tolerance and erreur_2 > tolerance :
    print(i)
    # 1. Calcul du delta T et du débit massique basé sur l'état actuel
    delta_t_1 = t_depart_consigne_1 - t_retour_estime_1
    cp_supply = fluid.get_heat_capacity(t_depart_consigne_1)
    cp_return = fluid.get_heat_capacity(t_retour_estime_1)
    M_dot_estimative_1=Heat_1/(((cp_supply+cp_return)/2)*delta_t_1)
    delta_t_2=t_depart_consigne_2 - t_retour_estime_2
    cp_supply_2 = fluid.get_heat_capacity(t_depart_consigne_2)
    cp_return_2 = fluid.get_heat_capacity(t_retour_estime_2)
    M_dot_estimative_2=Heat_2/(((cp_supply_2+cp_return_2)/2)*delta_t_2)
    # 2. Mise à jour du flow control dans le réseau
    net.circ_pump_mass.at[1,'mdot_flow_kg_per_s']=M_dot_estimative_1
    net.circ_pump_mass.at[2,'mdot_flow_kg_per_s']=M_dot_estimative_2
    # 3. Exécution du calcul du réseau
    pp.pipeflow(net, mode="bidirectional")
    # 4. Lecture de la VRAIE température de retour qui arrive à l'usine
    t_retour_reel_1 = net.res_circ_pump_mass.loc[1, "t_from_k"]
    t_retour_reel_2 = net.res_circ_pump_mass.loc[2, "t_from_k"]
    print(t_retour_reel_1,t_retour_reel_2)
    # 5. Calcul de l'erreur pour le critère d'arrêt
    erreur_1 = abs(t_retour_estime_1 - t_retour_reel_1)
    erreur_2 = abs(t_retour_estime_2 - t_retour_reel_2)
    print(erreur_1,erreur_2)
    # 6. Mise à jour pour la prochaine itération
    t_retour_estime_1 = t_retour_reel_1
    t_retour_estime_2 = t_retour_reel_2
    i=i+1
```
