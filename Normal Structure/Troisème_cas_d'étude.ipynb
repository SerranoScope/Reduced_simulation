```python
# =============================================================================
# COMPONENTE CONSUMIDOR (Potencia en W y Temp de Retorno)
# =============================================================================
# Conectamos el nudo de ida con el de retorno. 
# El sistema extraerá agua de la ida, absorberá los Watts indicados y la devolverá al retorno a la t_return_k.

#Consm1
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_1 ,
    to_junction=j_cons_ret_1,
    qext_w=150000,
    treturn_k=305,
    name="Consumidor_1"
)
#Consm2
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_2,
    to_junction=j_cons_ret_2,
    qext_w=150000,
    treturn_k=305,
    name="Consumidor_2"
)
#Consm3
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_3,
    to_junction=j_cons_ret_3,
    qext_w=150000,
    treturn_k=305,
    name="Consumidor_3"
)
#Consm4
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_4,
    to_junction=j_cons_ret_4,
    qext_w=150000,
    treturn_k=305,
    name="Consumidor_4"
)
#Consm5
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_5,
    to_junction=j_cons_ret_5,
    qext_w=150000,
    treturn_k=305,
    name="Consumidor_5"
)
#Consm6
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_6,
    to_junction=j_cons_ret_6,
    qext_w=150000,
    treturn_k=305,
    name="Consumidor_6"
)
```

```python
# =============================================================================
# CONDICIONES DE CONTORNO (EXT_GRID)
# =============================================================================
# Primeral Central 
pp.create_circ_pump_const_pressure(net,flow_junction=j_fuente_ida,return_junction=j_fuente_ret,p_flow_bar=3,plift_bar=0.5,t_flow_k=350 ,name='plant_1'
)
#Planta 2
fluid = get_fluid(net)
T_Supply=360
T_Return_estimative=305
cp_supply = fluid.get_heat_capacity(T_Supply)
cp_return = fluid.get_heat_capacity(T_Return_estimative)
Heat=200000
Delta_t_estimative=T_Supply-T_Return_estimative
M_dot_estimative=Heat/(((cp_supply+cp_return)/2)*Delta_t_estimative)

pp.create_circ_pump_const_mass_flow(
    net, return_junction=j_fuente_2_ret, flow_junction=j_fuente_2_inter,
    mdot_flow_kg_per_s=M_dot_estimative, t_flow_k=T_Supply, p_flow_bar=2.5  # valor arbitrario, ver nota abajo
)

pp.create_flow_control(
    net, from_junction=j_fuente_2_inter, to_junction=j_fuente_2_ida,
    controlled_mdot_kg_per_s=M_dot_estimative, loss_coefficient=0,
)
```

```python
tolerance = 0.0001  # Convergence à 0.01 °C près
erreur_1=10

t_depart_consigne_1 = 360 # 90°C en Kelvin
t_retour_estime_1 = 305 # Hypothèse de départ (60°C)

i=0
while erreur_1 > tolerance :
    
    # 1. Calcul du delta T et du débit massique basé sur l'état actuel
    delta_t_1 = t_depart_consigne_1 - t_retour_estime_1
    cp_supply = fluid.get_heat_capacity(t_depart_consigne_1)
    cp_return = fluid.get_heat_capacity(t_retour_estime_1)
    M_dot_estimative=Heat/(((cp_supply+cp_return)/2)*delta_t_1)
    # 2. Mise à jour du flow control dans le réseau
    net.flow_control.at[0,'controlled_mdot_kg_per_s']= M_dot_estimative
    net.circ_pump_mass.at[0,'mdot_flow_kg_per_s']=M_dot_estimative
    print(M_dot_estimative)
    # 3. Exécution du calcul du réseau
    pp.pipeflow(net, mode="bidirectional")
    # 4. Lecture de la VRAIE température de retour qui arrive à l'usine
    t_retour_reel_1 = net.res_junction.loc[19, "t_k"]
    # 5. Calcul de l'erreur pour le critère d'arrêt
    erreur_1 = abs(t_retour_estime_1 - t_retour_reel_1)
    # 6. Mise à jour pour la prochaine itération
    t_retour_estime_1 = t_retour_reel_1
    i=i+1
```
