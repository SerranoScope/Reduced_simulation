```python
import numpy as np
import pandapipes as pp

# 1. Crear red vacía
net = pp.create_empty_network(fluid="water")

# =============================================================================
# NUDOS (JUNCTIONS)
# =============================================================================
# Nota: Las temperaturas en pandapipes se introducen en Kelvin (K = °C + 273.15)
# Suponemos Ida a 80 °C (353.15 K) y Retorno esperado a 60 °C (333.15 K)


j_fuente_ida = pp.create_junction(net, pn_bar=5.0, tfluid_k=353.15, name="Fuente_Ida")
j_cons_ida = pp.create_junction(net, pn_bar=4.8, tfluid_k=353.15, name="Consumidor1_Ida")
j_cons_ida_2 = pp.create_junction(net, pn_bar=4.8, tfluid_k=353.15, name="Consumidor2_Ida")
j_cons_ida_3 = pp.create_junction(net, pn_bar=4.8, tfluid_k=353.15, name="Consumidor3_Ida")
j_cons_ida_4 = pp.create_junction(net, pn_bar=4.8, tfluid_k=353.15, name="Consumidor4_Ida")
j_fuente_ret = pp.create_junction(net, pn_bar=1.5, tfluid_k=333.15, name="Fuente_Retorno")
j_cons_ret = pp.create_junction(net, pn_bar=1.7, tfluid_k=333.15, name="Consumidor1_Retorno")
j_cons_ret_2 = pp.create_junction(net, pn_bar=1.7, tfluid_k=333.15, name="Consumidor2_Retorno")
j_cons_ret_3 = pp.create_junction(net, pn_bar=1.7, tfluid_k=333.15, name="Consumidor3_Retorno")
j_cons_ret_4 = pp.create_junction(net, pn_bar=1.7, tfluid_k=333.15, name="Consumidor4_Retorno")


# =============================================================================
# TUBERÍAS (PIPES) - IDA Y RETORNO SIN PÉRDIDAS TÉRMICAS (k_w = 0)
# =============================================================================

pipe_ida_1 = pp.create_pipe_from_parameters(
    net, from_junction=j_fuente_ida, to_junction=j_cons_ida,
    length_km=0.0155, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_Ida_Plant_Consm_1",
)

pipe_retorno_1= pp.create_pipe_from_parameters(
    net, from_junction=j_cons_ret, to_junction=j_fuente_ret,
    length_km=0.0155, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_RetornoPlant_Consm_1",
)

pipe_ida_2 = pp.create_pipe_from_parameters(
    net, from_junction=j_cons_ida, to_junction=j_cons_ida_2,
    length_km=0.0154, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_Ida_Consm1_Consm_2",
)
pipe_retorno_2= pp.create_pipe_from_parameters(
    net, from_junction=j_cons_ret_2, to_junction=j_cons_ret,
    length_km=0.0154, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_Retorno_Consm1_Consm_2",  
)
pipe_ida_3 = pp.create_pipe_from_parameters(
    net, from_junction=j_cons_ida_2, to_junction=j_cons_ida_3,
    length_km=0.0139, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_Ida_Consm2_Consm_3",
)

pipe_retorno_3= pp.create_pipe_from_parameters(
    net, from_junction=j_cons_ret_3, to_junction=j_cons_ret_2,
    length_km=0.0139, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_Retorno_Consm2_Consm_3",
)
pipe_ida_4 = pp.create_pipe_from_parameters(
    net, from_junction=j_cons_ida_3, to_junction=j_cons_ida_4,
    length_km=0.0176, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_Ida_Consm3_Consm_4",
)

pipe_retorno_4= pp.create_pipe_from_parameters(
    net, from_junction=j_cons_ret_4, to_junction=j_cons_ret_3,
    length_km=0.0176, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_Retorno_Consm3_Consm_4",
)
pipe_ida_5= pp.create_pipe_from_parameters(
    net, from_junction=j_cons_ida_4, to_junction=j_fuente_ida,
    length_km=0.0128, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_ida_Consm4_plant",
)
pipe_retorno_5= pp.create_pipe_from_parameters(
    net, from_junction=j_fuente_ret, to_junction=j_cons_ret_4,
    length_km=0.0128, inner_diameter_mm=150, u_w_per_m2k=0, k_mm=0.1*1000, text_k=273.2, name="Tubo_Retorno_Consm4_plant",
)
# =============================================================================
# COMPONENTE CONSUMIDOR (Potencia en W y Temp de Retorno)
# =============================================================================
# Conectamos el nudo de ida con el de retorno. 
# El sistema extraerá agua de la ida, absorberá los Watts indicados y la devolverá al retorno a la t_return_k.

pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida,
    to_junction=j_cons_ret,
    qext_w=300000,
    treturn_k=313.2,
    name="Consumidor_1"
)
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_2,
    to_junction=j_cons_ret_2,
    qext_w=300000,
    treturn_k=313.2,
    name="Consumidor_2"
)
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_3,
    to_junction=j_cons_ret_3,
    qext_w=300000,
    treturn_k=313.2,
    name="Consumidor_3"
)
pp.create_heat_consumer(
    net,
    from_junction=j_cons_ida_4,
    to_junction=j_cons_ret_4,
    qext_w=300000,
    treturn_k=313.2,
    name="Consumidor_4"
)

# =============================================================================
# CONDICIONES DE CONTORNO (EXT_GRID)
# =============================================================================
# Fijamos la presión y temperatura de salida en la ida

pp.create_circ_pump_const_pressure(net,flow_junction=j_fuente_ida,return_junction=j_fuente_ret,p_flow_bar=3,plift_bar=0.5,t_flow_k=353 ,name='Grid'
)


# =============================================================================
# SIMULACIÓN Y RESULTADOS
# =============================================================================
pp.pipeflow(net, mode="bidirectional")
```
